**Objective:** Visualize patterns in the Iris dataset using a bar chart, a histogram and a scatter plot, and compare its features.

Loaded the Iris dataset, which has 150 flowers and 5 columns: four measurements (sepal and petal length and width) and the species.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("iris.csv")
print(df.shape)
df.head()
```

Checked the data types and missing values. There are none, and each of the 3 species (setosa, versicolor, virginica) has 50 flowers.

```python
df.info()
df["species"].value_counts()
```

Used describe() to see the average, minimum and maximum of each measurement before charting.

```python
df.describe()
```

```python
avg_petal = df.groupby("species")["petal_length"].mean()

avg_petal.plot(kind="bar", color=["#4C72B0", "#55A868","#C42E52"], figsize=(6,4))
plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.xticks(rotation=0)
plt.show()
```

```python
df["petal_length"].plot(kind="hist", bins=20, edgecolor="black", figsize=(6,4))
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Number of Flowers")
plt.show()
```

```python
sns.scatterplot(data=df, x="petal_length",y="petal_width", hue="species")
plt.title("Petal Length VS Petal Width by Species")
plt.xlabel("Petal Legth (cm)")
plt.ylabel("Petal Widt (cm)")
plt.show()
```

```python
df.groupby("species").mean().round(2)
```

```python
sns.heatmap(df.drop(columns="species").corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Between Features")
plt.show()
```

```python
sns.pairplot(df, hue="species")
plt.show()
```

## Key Insights
- Setosa is clearly different from the other two species, mainly because of its small petals.
- Petal length and petal width are the best features for telling the species apart.
- Versicolor and virginica overlap a little, so they are harder to separate.
- Petal length and width are strongly correlated (0.96), while sepal width has a weak negative relationship with the other features.
